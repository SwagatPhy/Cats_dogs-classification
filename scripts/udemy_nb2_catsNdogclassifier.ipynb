```python
# load the train and test files 
cat_train = (glob(f'{train_cat_dir}/*'))
dog_train = (glob(f'{train_dog_dir}/*'))


cat_test = (glob(f'{test_cat_dir}/*'))
dog_test = (glob(f'{test_dog_dir}/*'))

num_files_train = len(cat_train)


train_cat_features = []
train_dog_features = []
test_cat_features = []
test_dog_features = []

for i in range(0,num_files_train):

    train_cat_image = cv2.imread(cat_train[i])
    train_cat_resize_img = cv2.resize(train_cat_image, (200,200), interpolation=cv2.INTER_AREA)
    train_cat_features.append(train_cat_resize_img)

    train_dog_image = cv2.imread(dog_train[i])
    train_dog_resize_img = cv2.resize(train_dog_image, (200,200), interpolation=cv2.INTER_AREA)
    train_dog_features.append(train_dog_resize_img)




    ## view file
    # cv2.imshow('webp image', train_cat_resize)
    # cv2.waitKey(0)
    # cv2.destroyAllWindows()
    print(train_cat_resize_img.shape)



num_files_test = len(cat_test)
for j in range(0, num_files_test):

    test_cat_image = cv2.imread(cat_test[j])
    test_cat_resize_img = cv2.resize(test_cat_image, (200,200), interpolation=cv2.INTER_AREA)
    test_cat_features.append(test_cat_resize_img)

    test_dog_image = cv2.imread(dog_test[j])
    test_dog_resize_img = cv2.resize(test_dog_image, (200,200), interpolation=cv2.INTER_AREA)
    test_dog_features.append(test_dog_resize_img)


    # cv2.imshow('webp image', train_cat_resize)
    # cv2.waitKey(0)
    # cv2.destroyAllWindows()
    print(test_cat_resize_img.shape)
```

```python
import tensorflow as tf

# model = tf.keras.Sequential([
#     # tf.keras.Input(shape=(200,200,3)),
#     tf.keras.layers.Flatten(input_shape=(200,200,3)),
#     tf.keras.layers.Dense(units=32, activation='relu'),
#     tf.keras.layers.Dense(units=64, activation='relu'),
#     tf.keras.layers.Dense(units=1, activation='sigmoid')
# ])

model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=(200,200,3)),
    tf.keras.layers.MaxPool2D(2,2),
    tf.keras.layers.Conv2D(64, kernel_size=(3,3), activation='relu'),
    tf.keras.layers.MaxPool2D(2,2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128,  activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model_history = model.fit(x_train, y_train, batch_size=32, epochs=10, validation_data=(x_test, y_test))
```
